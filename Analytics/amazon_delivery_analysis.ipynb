```python
df.groupby("Traffic")["Delivery_Time"].mean()

# Problem 1: Traffic congestion significantly increases delivery time.
# Evidence: Deliveries during Jam traffic take an average of 147.76 minutes, compared with 101.35 minutes during Low traffic — a difference of approximately 46.4 minutes.
# Business Impact: Severe traffic congestion can lead to delayed deliveries and negatively affect customer satisfaction.
# Recommendation: The company could prioritize route optimization and assign delivery agents based on traffic conditions to reduce delays.
# Data Quality Issue: 91 delivery records have missing Traffic and Weather information, which limits analysis of environmental factors.
```

```python
df.groupby("Weather")["Delivery_Time"].mean()
# Problem : Poor weather conditions are associated with longer delivery times.
# Evidence : Deliveries during Cloudy weather have the highest average delivery time at 138.29 minutes, while Sunny conditions have the lowest at 103.66 minutes. This represents a difference of approximately 34.62 minutes.
# Business Impact : Weather conditions may slow down delivery operations and increase the likelihood of delayed orders.
# Recommendation : The company could allocate additional delivery time during unfavorable weather conditions and prioritize safer/faster delivery routes.
```

```python
df["Traffic"].isna().sum()
df["Traffic"].unique()
df["Traffic"] = df["Traffic"].replace( "NaN" , np.nan)
```

```python
traffic_df = df.dropna(subset=["Traffic"])
```

```python
sns.boxplot(x="Traffic", y="Delivery_Time", data=traffic_df)

plt.xlabel("Traffic Type")
plt.ylabel("Delivery Time (minutes)")
plt.title("Delivery Time by Traffic Condition")
plt.show()
```

```python
traffic_df = df[
    df["Traffic"].notna() &
    (df["Traffic"].astype(str).str.strip().str.lower() != "nan")
]

sns.boxplot(x="Traffic", y="Delivery_Time", data=traffic_df)

plt.xlabel("Traffic Type")
plt.ylabel("Delivery Time (minutes)")
plt.title("Delivery Time by Traffic Condition")
plt.show()
```

```python
sns.boxplot(x = "Weather" , y = "Delivery_Time" , data = df)
plt.xlabel("Weather Condition")
plt.ylabel("Delivery Time (minutes)")
plt.title("Delivery Time By Weather Condititon")
plt.show()
```

```python
sns.boxplot( x = "Vehicle" , y = "Delivery_Time" , data = df)
plt.xlabel("Vehicle Type")
plt.ylabel("Delivery_Time (minutes)")
plt.title("Delivery Time took Vehicle to Deliver")
plt.show()
```

```python
sns.barplot(x="Delivery_Time", y="Category", data=df , errorbar = None)

plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Category")
plt.title("Average Delivery Time by Category")

plt.show()
```

```python
sns.boxplot(x = "Area" , y = "Delivery_Time" , data = df)
plt.xlabel("Delivery Area")
plt.ylabel("Delivery Time")
plt.title("Delivery Time By Delivery Area")
plt.show()
```

```python
traffic_weather = df.groupby(["Traffic" , "Weather"])["Delivery_Time"].mean().unstack()
```
